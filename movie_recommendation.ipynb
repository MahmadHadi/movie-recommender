```python
import numpy as np
import pandas as pd
```

```python
movies = pd.read_csv('/kaggle/input/tmdb-movie-metadata/tmdb_5000_movies.csv')
credits = pd.read_csv('/kaggle/input/tmdb-movie-metadata/tmdb_5000_credits.csv')
```

```python
movies.head(3)
```

```python
credits.head(3)
```

```python
# merge both dataframes -> movies & credits
movies = movies.merge(credits, on='title')
```

```python
movies.head(1).columns
```

```python
# remove unnecessary columns
# since we are developing an content based movie recommendation system, we need to generate tags and by comparing tags, we will recommend movies
# so before selecting columns, we have to ask "will it help me to create tags ?"

cols = ['genres', 'keywords', 'overview', 'title', 'cast', 'crew', 'movie_id']
movies = movies[cols]
movies.head(3)
```

```python
# check for missing values
movies.isna().sum()
movies = movies.dropna() # because only 3 missing values in overview
movies.isna().sum()
```

```python
# check for duplicate data
movies.duplicated().sum() # no duplicate
```

```python
# format genres -> json format to normal list
# input: [{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}, {"id": 14, "name": "Fantasy"}, {"id": 878, "name": "Science Fiction"}]
# output: ['Action', 'Adventure', 'Fantasy']
# movies.iloc[0].genres

# USING numpy -> good for large data
import json
def convert(obj):
    data = json.loads(obj)
    return np.array([item['name'] for item in data]).tolist()
```

```python
movies['genres'] = movies['genres'].apply(convert)
```

```python
movies['keywords'] = movies['keywords'].apply(convert)
```

```python
# get top 3 actors from the cast
import json
def convert_cast(obj):
  # obj -> json string
  obj = json.loads(obj)[:3]
  li = []
  for i in obj:
    li.append(i['name'])
  return li
```

```python
movies['cast'] = movies['cast'].apply(convert_cast)
```

```python
# now movies crew
# we only need those crew where the job == 'director'
import json
def convert_crew(obj):
  obj = json.loads(obj)
  return [u['name'] for u in obj if u['job'].lower() == 'director']
```

```python
movies['crew'] = movies['crew'].apply(convert_crew)
```

```python
# convert overview string -> list
movies['overview'] = movies['overview'].apply(lambda x: x.split())
```

```python
# Remove spaces between words.
# Names like 'mahmad hadi' and 'mahmad ali' share the token 'mahmad',
# which can confuse the model into treating them as the same entity.
# So we merge them into single tokens:
#   'mahmad hadi' -> 'mahmadhadi'
#   'mahmad ali'  -> 'mahmadali'

movies['genres'] = movies['genres'].apply(lambda x: [i.replace(" ", "") for i in x])
movies['keywords'] = movies['keywords'].apply(lambda x: [i.replace(" ", "") for i in x])
movies['overview'] = movies['overview'].apply(lambda x: [i.replace(" ", "") for i in x])
movies['cast'] = movies['cast'].apply(lambda x: [i.replace(" ", "") for i in x])
movies['crew'] = movies['crew'].apply(lambda x: [i.replace(" ", "") for i in x])
```

```python
# now create a new col named tags wich will be merged of all other colums
movies['tags'] = movies['genres'] + movies['keywords'] + movies['overview'] + movies['cast'] + movies['crew']
```

```python
movies.head(3)
```

```python
new_df = movies[['title', 'tags', 'movie_id']]
new_df.head(3)
```

```python
# conert tags list into string

new_df['tags'] = new_df['tags'].apply(lambda x: " ".join(x).lower())
```

```python
new_df.head(3)
```

```python
import nltk
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(tag):
  tag = tag.split()

  tag = [ps.stem(i) for i in tag]
  return " ".join(tag)

new_df['tags'] = new_df['tags'].apply(stem)
```

# Vectorization, Stop Words & Bag of Words (for tag strings)

### Vectorization
Converting text (tag strings) into numeric vectors so a model can process it. Each unique tag becomes a dimension/feature. Example: `"action comedy"` → `[1, 1, 0]` over vocabulary `[action, comedy, drama]`.

### Stop Words
Common words with little meaning (e.g., `the`, `is`, `and`, `of`) that are removed before vectorizing. This reduces noise and shrinks the vocabulary, so the model focuses on meaningful tags.

### Bag of Words (BoW)
A vectorization method that represents text as **word counts**, ignoring order. For tags:
- `"action action comedy"` → `{action: 2, comedy: 1}`
- Used to turn each movie's tags into a fixed-length numeric vector for similarity/search.

**Pipeline for tag strings:**
`raw tags → remove stop words → tokenize → BoW vectorize → numeric vector`

```python
from sklearn.feature_extraction.text import CountVectorizer

# max_features -> how many words in vector (vocabulary size cap)
#   keeps only the top N most frequent words; rest ignored; here N = 5000

cv = CountVectorizer(max_features=5000, stop_words='english')
```

```python
vectors = cv.fit_transform(new_df['tags']).toarray()
```

# Cosine Similarity & Distance — Movie Recommendation

- **What:** measures angle between two vectors (movie tag vectors), ignoring magnitude.
- **Similarity:** `1` = identical direction, `0` = unrelated, `-1` = opposite.
- **Distance:** `1 - similarity`; `0` = same, larger = more different.
- **Why here:** each movie is a BoW/TF-IDF vector → pick movies with highest similarity to the query movie.
- **Formula:** `cos(θ) = (A·B) / (‖A‖‖B‖)`
- **Watch out:**
  - Uses **direction, not length** — a short and long tag list can still be similar.
  - Works on **sparse vectors** (sklearn returns a sparse matrix).
  - `pairwise_distances` gives distance; use `1 - distance` or `cosine_similarity` for scores.
- **Code:** `from sklearn.metrics.pairwise import cosine_similarity`

```python
from sklearn.metrics.pairwise import cosine_similarity

# compares every movie with every other movie
# similarity[i][j] = how alike movie i and movie j are (1 = same, 0 = unrelated)
similarity = cosine_similarity(vectors)
```

```python
def recommend(movie):
  movie_index = new_df[new_df['title'].apply(lambda x: x.lower()) == movie.lower()].index[0]

  distances = similarity[movie_index]
  movies_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x: x[1])[1:6]

  for i in movies_list:
    print(new_df.iloc[i[0]].title)
  print()
```

```python
recommend("Avatar")
recommend("The Dark Knight")
recommend("Toy Story")
```
